# Introduction

Healthcare quality is a critical component of modern healthcare systems, influencing patient outcomes, healthcare costs, and overall public trust. Among the many indicators used to evaluate hospital performance, hospital readmission rates have received considerable attention because they reflect the effectiveness of clinical care, discharge planning, and care coordination. Excessive readmissions not only increase healthcare expenditures but may also indicate opportunities for improving patient care.


This project investigates the relationships between organizational characteristics, patient experience, and hospital quality outcomes across U.S. hospitals. By integrating multiple CMS datasets, the analysis explores how hospital ratings, patient experience, ownership, emergency service availability, and clinical quality measures are associated with excess readmission performance. Interactive visualizations and statistical analyses are used to identify patterns, compare hospital performance, and provide insights into the factors associated with high-quality healthcare delivery.

# Data Collection

The data used in this project were obtained from the **Centers for Medicare & Medicaid Services (CMS) Care Compare** public reporting program. CMS publishes hospital quality datasets that evaluate multiple aspects of healthcare performance, including patient experience, hospital characteristics, clinical outcomes, safety, and readmission measures.

Three publicly available CMS datasets were combined to create a comprehensive hospital quality dataset:

1. **Hospital General Information**
   - Hospital identifiers and names
   - State
   - Hospital ownership
   - Hospital type
   - Emergency service availability
   - CMS Overall Hospital Rating
   - Summary counts of mortality, safety, and readmission measures performing better than the national average

2. **Hospital Consumer Assessment of Healthcare Providers and Systems (HCAHPS)**
   - Nurse communication
   - Doctor communication
   - Communication about medicines
   - Care transition
   - Cleanliness
   - Quietness
   - Overall patient rating
   - Likelihood of recommending the hospital

   These measures were combined to construct a **Patient Experience Index**, representing overall patient-centered care performance.

3. **Hospital Readmission Reduction Program**
   - Excess Readmission Ratio
   - Predicted Readmission Rate
   - Expected Readmission Rate
   - Number of Discharges
   - Number of Readmissions
   - Clinical readmission measures for six major conditions and procedures

The datasets were integrated using the unique **Facility ID**, producing a unified dataset that combines organizational characteristics, patient experience measures, and clinical quality indicators for hospitals across the United States.

# Research Question 3

## Does hospital ownership influence excess readmission performance?

### Background

Hospital ownership influences organizational structure, funding models, and operational priorities, which may affect healthcare quality and patient outcomes. Comparing readmission performance across ownership types helps determine whether ownership is associated with differences in hospital quality.

### Objective

To compare excess readmission ratios among different hospital ownership categories.

```python
# @title
rq3 = final_df[
    [
        "Hospital Ownership",
        "Excess Readmission Ratio"
    ]
].copy()

rq3 = rq3.dropna()

print(rq3.shape)
rq3.head()
```

```python
# @title
counts = rq3["Hospital Ownership"].value_counts()

valid_groups = counts[counts >= 100].index

rq3 = rq3[rq3["Hospital Ownership"].isin(valid_groups)]
```

```python
ownership_summary = (
    rq3.groupby("Hospital Ownership")["Excess Readmission Ratio"]
       .agg(
           Mean="mean",
           Median="median",
           Std="std",
           Count="count"
       )
       .round(3)
       .sort_values("Mean")
)

ownership_summary
```

```python
import plotly.express as px

fig = px.box(
    rq3,
    x="Hospital Ownership",
    y="Excess Readmission Ratio",
    color="Hospital Ownership",
    points="outliers",
    title="Excess Readmission Ratio by Hospital Ownership"
)

fig.update_layout(
    template="plotly_white",
    height=600,
    title_x=0.5,
    showlegend=False,
    xaxis_title="Hospital Ownership",
    yaxis_title="Excess Readmission Ratio"
)

fig.update_xaxes(tickangle=20)

fig.show()
```

```python
from scipy.stats import f_oneway

groups = [
    group["Excess Readmission Ratio"].values
    for _, group in rq3.groupby("Hospital Ownership")
]

f_stat, p_value = f_oneway(*groups)

print(f"F-statistic: {f_stat:.3f}")
print(f"P-value: {p_value:.4f}")
```

### Interpretation

The box plot and summary statistics compare excess readmission ratios across major hospital ownership categories.

Among the ownership groups, **Voluntary Non-profit (Church)** hospitals achieved the lowest average Excess Readmission Ratio (0.99), indicating slightly better readmission performance. In contrast, **Proprietary** hospitals recorded the highest average ratio (1.02), while the remaining government and non-profit ownership groups clustered around the overall average of 1.00.

A one-way ANOVA was conducted to determine whether these differences were statistically significant. The analysis produced an **F-statistic of 15.255** with a **p-value < 0.001**, indicating statistically significant differences in excess readmission ratios across hospital ownership categories.

Although the differences in average readmission ratios are relatively small, the results suggest that ownership structure is associated with hospital readmission performance. However, ownership alone is unlikely to explain these differences, as hospital resources, patient populations, and organizational practices also contribute to readmission outcomes.

### Key Takeaway

Hospital ownership is significantly associated with excess readmission performance. Non-profit hospitals generally demonstrated slightly lower readmission ratios than proprietary hospitals, although the magnitude of the differences was modest, suggesting that multiple organizational and clinical factors influence hospital outcomes.

# Research Question 4

## Do hospitals with emergency services provide a better patient experience than hospitals without emergency services?

### Background

Emergency departments are often the primary point of contact for patients requiring urgent medical care. Hospitals providing emergency services generally manage a higher volume and greater complexity of patients, which may influence patients' overall healthcare experience.

### Objective

To compare the Patient Experience Index between hospitals that provide emergency services and those that do not.

```python
# @title
rq4 = final_df[
    [
        "Emergency Services",
        "Patient Experience Index"
    ]
].copy()

rq4 = rq4.dropna()

rq4.head()
```

```python
summary = (
    rq4.groupby("Emergency Services")["Patient Experience Index"]
       .agg(
           Mean="mean",
           Median="median",
           Std="std",
           Count="count"
       )
       .round(2)
)

summary
```

```python
import plotly.express as px

fig = px.violin(
    rq4,
    x="Emergency Services",
    y="Patient Experience Index",
    color="Emergency Services",
    box=True,
    points="outliers",
    title="Patient Experience Index by Emergency Service Availability"
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5,
    showlegend=False,
    height=600,
    xaxis_title="Emergency Services",
    yaxis_title="Patient Experience Index"
)

fig.show()
```

```python
from scipy.stats import ttest_ind

yes = rq4[rq4["Emergency Services"]=="Yes"]["Patient Experience Index"]

no = rq4[rq4["Emergency Services"]=="No"]["Patient Experience Index"]

t_stat, p_value = ttest_ind(
    yes,
    no,
    equal_var=False
)

print(f"T-statistic: {t_stat:.3f}")
print(f"P-value: {p_value:.4f}")
```

### Interpretation

The comparison of patient experience scores indicates a statistically significant difference between hospitals that provide emergency services and those that do not.

Hospitals **without emergency services** reported a higher average Patient Experience Index (**86.97**) than hospitals **with emergency services** (**85.18**). Although the difference is relatively small (approximately 1.8 points), it was statistically significant according to the independent samples t-test (**t = -15.120, p < 0.001**).

One possible explanation is that hospitals offering emergency services often manage higher patient volumes, more urgent medical conditions, and greater operational complexity. These factors may influence patients' perceptions of communication, waiting times, and overall care experience.

Overall, the findings suggest that the availability of emergency services is associated with differences in patient experience, although this relationship should be interpreted within the broader context of hospital size, case complexity, and service demands.

### Key Takeaway

Hospitals without emergency services achieved significantly higher patient experience scores than hospitals with emergency services. This suggests that operational complexity and patient case mix may influence patient satisfaction in addition to the quality of care delivered.

# Research Question 8

## Which hospitals demonstrate balanced performance across multiple quality dimensions?

### Background

Hospital quality extends beyond a single performance indicator. High-performing hospitals are expected to provide excellent patient experiences, maintain strong clinical outcomes, ensure patient safety, and minimize avoidable readmissions. Evaluating these dimensions together provides a more comprehensive understanding of hospital performance than examining each indicator individually.

### Objective

To develop a composite quality score by combining multiple quality dimensions and identify hospitals that demonstrate consistently strong performance across patient experience, clinical quality, safety, and readmission outcomes.

### Quality Dimensions Included

The composite quality assessment incorporates the following indicators:

- Hospital Overall Rating
- Patient Experience Index
- Excess Readmission Ratio
- Count of Mortality Measures Better than the National Average
- Count of Safety Measures Better than the National Average
- Count of Readmission Measures Better than the National Average

Since these indicators are measured on different scales, they will be standardized before calculating an overall quality score. Hospitals will then be ranked to identify those demonstrating the most balanced performance across all quality dimensions.

```python
import pandas as pd
import numpy as np

rq8 = final_df[
    [
        "Facility ID",
        "Facility Name",
        "State",
        "Hospital overall rating",
        "Patient Experience Index",
        "Excess Readmission Ratio",
        "Count of MORT Measures Better",
        "Count of Safety Measures Better",
        "Count of READM Measures Better"
    ]
].copy()

rq8 = rq8.dropna()

rq8.head()
```

```python
hospital_scores = (
    rq8.groupby(["Facility ID", "Facility Name", "State"], as_index=False)
       .agg({
           "Hospital overall rating": "mean",
           "Patient Experience Index": "mean",
           "Excess Readmission Ratio": "mean",
           "Count of MORT Measures Better": "mean",
           "Count of Safety Measures Better": "mean",
           "Count of READM Measures Better": "mean"
       })
)
```

```python
quality_cols = [
    "Hospital overall rating",
    "Patient Experience Index",
    "Count of MORT Measures Better",
    "Count of Safety Measures Better",
    "Count of READM Measures Better"
]

for col in quality_cols:
    hospital_scores[col + "_z"] = (
        hospital_scores[col] - hospital_scores[col].mean()
    ) / hospital_scores[col].std()

hospital_scores["Readmission_z"] = -(
    (hospital_scores["Excess Readmission Ratio"] -
     hospital_scores["Excess Readmission Ratio"].mean())
    / hospital_scores["Excess Readmission Ratio"].std()
)
```

```python
hospital_scores["Composite Quality Score"] = hospital_scores[
    [
        "Hospital overall rating_z",
        "Patient Experience Index_z",
        "Count of MORT Measures Better_z",
        "Count of Safety Measures Better_z",
        "Count of READM Measures Better_z",
        "Readmission_z"
    ]
].mean(axis=1)
```

```python
top_hospitals = (
    hospital_scores[
        [
            "Facility Name",
            "State",
            "Composite Quality Score"
        ]
    ]
    .sort_values("Composite Quality Score", ascending=False)
    .head(10)
)

top_hospitals
```

```python
# @title
quality_cols = [
    "Hospital overall rating",
    "Patient Experience Index",
    "Count of MORT Measures Better",
    "Count of Safety Measures Better",
    "Count of READM Measures Better"
]

for col in quality_cols:
    rq8[col + "_z"] = (
        rq8[col] - rq8[col].mean()
    ) / rq8[col].std()

# Lower readmission ratio is better, so reverse the sign
rq8["Readmission_z"] = -(
    (rq8["Excess Readmission Ratio"] -
     rq8["Excess Readmission Ratio"].mean())
    / rq8["Excess Readmission Ratio"].std()
)
```

### Interpretation

A composite quality score was developed by combining standardized measures of hospital overall rating, patient experience, mortality performance, safety performance, readmission performance, and excess readmission ratios. Hospitals were then ranked based on their overall performance across these quality dimensions.

The results identified **NYU Langone Hospitals (New York)** as the highest-performing hospital, achieving the highest composite quality score of **3.94**. Other leading hospitals included **Mayo Clinic Hospital Rochester (Minnesota)**, **Massachusetts General Hospital (Massachusetts)**, **Morristown Medical Center (New Jersey)**, and **Cedars-Sinai Medical Center (California)**.

These hospitals consistently performed well across multiple quality dimensions, demonstrating strong patient experience, superior clinical outcomes, better safety performance, and lower excess readmission ratios. Rather than excelling in a single metric, the highest-ranked hospitals maintained consistently high performance across all indicators included in the composite quality score.

Overall, the findings highlight that excellence in hospital quality is multidimensional, requiring strong performance across patient-centered care, clinical quality, patient safety, and readmission outcomes.

### Key Takeaway

The highest-ranked hospitals demonstrated consistently strong performance across multiple quality dimensions rather than excelling in only one area. This suggests that delivering high-quality healthcare requires a balanced approach that combines excellent patient experience, strong clinical outcomes, effective safety practices, and reduced hospital readmissions.
